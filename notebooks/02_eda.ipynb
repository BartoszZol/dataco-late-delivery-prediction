```python
df['shipping_mode']=df['shipping_mode'].astype('category')
df['payment_type']=df['payment_type'].astype('category')
df['customer_segment']=df['customer_segment'].astype('category')
df['market']=df['market'].astype('category')
```

```python
categorical_features = [
    'shipping_mode', 'payment_type', 'customer_segment',
    'market', 'department_name', 'order_region',
    'category_name', 'order_status'
]

for col in categorical_features:
    print(df.groupby(col, observed=True)['late_delivery_risk'].mean().sort_values(ascending=False))
    print()
```
